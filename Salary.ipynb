```python
import pandas as pd

data = pd.read_csv('Salary_Data.csv') 
data.head()
```

```python
import seaborn as sns

sns.scatterplot(data = data, x = 'YearsExperience', y = 'Salary', marker = '*', color = 'red')
```

```python
# Prepare data 
# Separate input and output

X = data[['YearsExperience']]   # input must be 2D
y = data['Salary']
```

```python
data.shape
```

```python
# Load Algorithm
# For Machine Learning, we use scikit-learn library. It has many algorithms for machine learning. We will use Linear Regression algorithm for this problem.

from sklearn.linear_model import LinearRegression
```

```python
# To get access to the members of the LinearRegression class, we need to create an object of the class. We will call this object as model.
model = LinearRegression()
```

```python
model
```

```python
# Train the model
model.fit(X, y)
```

```python
# Slop
model.coef_
```

```python
model.intercept_
```

```python
# Training/Prediction --> Predict
import warnings
warnings.filterwarnings('ignore')

model.predict([[.6]])
```

```python
model.predict([[.6],[4],[0],[10]])
```

```python
model.predict([[0]])
```

```python
# y = m*x + c
model.coef_ * 0 + model.intercept_
```

```python
model.coef_ * 4 + model.intercept_
```

```python
model.coef_ * 6 + model.intercept_
```

```python
test_data = [[0.2], [3], [6], [10], [12]]
y_pred = model.predict(test_data)
y_pred
```

```python
import matplotlib.pyplot as plt
plt.plot(test_data, y_pred, color = 'red', label = 'Predicted Salary')
plt.scatter(X, y, color = 'm', label = 'Actual Data')
plt.legend()
plt.show()
```
